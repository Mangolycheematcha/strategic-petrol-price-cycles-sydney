This notebook calls the reusable `src` workflow, then displays the generated audit tables, robustness comparison, cycle metrics and figures.

```python
from pathlib import Path
import sys


def find_project_root(start: Path) -> Path:
    for folder in [start, *start.parents]:
        if (folder / "data").exists() and (folder / "notebooks").exists() and (folder / "src").exists():
            return folder
    raise FileNotFoundError("Could not locate the repository root.")


PROJECT_ROOT = find_project_root(Path.cwd())
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print(f"Project root: {PROJECT_ROOT}")
```

```python
from src.run_analysis import run

results = run()
print(f"Source months: {results['manifest']['covered_period'].min()} to {results['manifest']['covered_period'].max()}")
print(f"Raw rows loaded: {results['raw_rows']:,}")
print(f"Clean U91 Sydney buffer+analysis events: {results['clean_events']:,}")
print(f"Primary complete cycles detected: {results['selected_cycle_count']}")
```
